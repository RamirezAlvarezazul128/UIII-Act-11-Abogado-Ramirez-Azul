### Proyecto de simulación de inteligencia artificial  
### Problema 48 Menu Abogado
Especialidad (penal/civil), bufete, casos ganados, número de cédula, juzgados frecuentes y honorarios.

```python
# --- Programa de Gestión y Recomendación de Abogados (Simulación IA) ---

# Iniciamos con una lista vacía para almacenar los diccionarios de abogados [1, 4, 5]
base_datos_abogados = []

def mostrar_menu():
    print("\n" + "="*50)
    print("      SISTEMA DE ASISTENCIA LEGAL (IA)")
    print("="*50)
    print("1. Registrar nuevo abogado")
    print("2. Ver listado de abogados (Tabla)")
    print("3. Consultar Asistente IA (Recomendación)")
    print("4. Salir")
    print("="*50)

def registrar_abogado():
    print("\n--- Registro de Datos ---")
    # Creamos un diccionario vacío para el nuevo abogado y solicitamos datos [1, 6]
    abogado = {}

    abogado['Nombre'] = input("Nombre completo: ")
    abogado['Especialidad'] = input("Especialidad (penal/civil): ").lower()
    abogado['Bufete'] = input("Nombre del bufete: ")
    abogado['Casos Ganados'] = int(input("Número de casos ganados: "))
    abogado['Cédula'] = input("Número de cédula profesional: ")
    abogado['Juzgados'] = input("Juzgados frecuentes: ")
    abogado['Honorarios'] = float(input("Honorarios por consulta ($): "))

    # Agregamos el diccionario a nuestra lista global [7]
    base_datos_abogados.append(abogado)
    print("\n¡Abogado registrado exitosamente!")

def mostrar_tabla():
    if not base_datos_abogados:
        print("\nNo hay abogados registrados en el sistema.")
    else:
        # Encabezado de la tabla alineado con f-strings [8]
        print("\n" + "-"*110)
        print(f"{'Nombre':<15} | {'Especialidad':<12} | {'Bufete':<15} | {'Casos':<6} | {'Cédula':<10} | {'Honorarios':<10}")
        print("-"*110)

        # Ciclo for para recorrer la lista de diccionarios [1, 9]
        for abog in base_datos_abogados:
            print(f"{abog['Nombre']:<15} | {abog['Especialidad']:<12} | {abog['Bufete']:<15} | "
                  f"{abog['Casos Ganados']:<6} | {abog['Cédula']:<10} | ${abog['Honorarios']:<10.2f}")
        print("-"*110)

def asistente_ia():
    print("\n--- Asistente Virtual IA ---")
    if not base_datos_abogados:
        print("La IA no tiene datos suficientes para recomendar.")
        return

    print("IA: ¿Qué tipo de problema legal tiene (penal/civil)?")
    necesidad = input("Usuario: ").lower()

    print("IA: ¿Cuál es su presupuesto máximo?")
    presupuesto = float(input("Usuario: "))

    encontrado = False
    print("\nIA: Analizando perfiles y recomendando...")

    for abog in base_datos_abogados:
        # Lógica de decisión de la "IA" utilizando operadores relacionales y lógicos [1, 10]
        if abog['Especialidad'] == necesidad and abog['Honorarios'] <= presupuesto:
            print(f"\n> RECOMENDACIÓN ENCONTRADA:")
            print(f"El abogado {abog['Nombre']} es experto en el área {necesidad}.")
            print(f"Tiene {abog['Casos Ganados']} casos ganados y cobra ${abog['Honorarios']}.")

            # Sub-decisión basada en el éxito del abogado [11]
            if abog['Casos Ganados'] > 50:
                print("Nota de la IA: Este perfil tiene un alto índice de éxito.")
            elif abog['Casos Ganados'] > 20:
                print("Nota de la IA: Perfil con experiencia sólida.")
            else:
                print("Nota de la IA: Profesional con honorarios competitivos.")

            encontrado = True
            break # Detenemos la búsqueda al encontrar el primero óptimo [12]

    if not encontrado:
        print("IA: Lo siento, no encontré un perfil que se ajuste exactamente a sus filtros.")

# --- Bucle Principal del Programa [13] ---
while True:
    mostrar_menu()
    opcion = input("Seleccione una opción: ")

    # Estructura condicional if-elif-else para el menú [1, 10]
    if opcion == "1":
        registrar_abogado()
    elif opcion == "2":
        mostrar_tabla()
    elif opcion == "3":
        asistente_ia()
    elif opcion == "4":
        print("Cerrando sistema de asistencia... ¡Hasta pronto!")
        break # Sale del bucle True para finalizar el programa
    else:
        print("Opción no válida. Por favor, intente de nuevo.")
```
